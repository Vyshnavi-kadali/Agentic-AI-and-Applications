```python
#2. Tokenization Demo
sentence = "LLMs are powerful tools for natural language understanding."

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("distilgpt2")

tokens = tokenizer.tokenize(sentence)
token_ids = tokenizer.convert_tokens_to_ids(tokens)
sequence_length = len(token_ids)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
print("Sequence Length:", sequence_length)
```

```python
#Explaination for Tokenisation

#Tokenization is an important step in Natural Language Processing (NLP). It refers to the process of converting raw text into smaller units called tokens, which are later converted into numerical values that machine learning models can understand. 

#Since computers cannot directly process human language, text must be transformed into a numerical format before being given to a model. During tokenization, an input sentence is split into tokens that may represent full words, parts of words (subwords), or punctuation marks. Most modern language models use subword tokenization, which helps in handling new or rare words efficiently.

#Each token is then mapped to a unique numerical identifier using a predefined vocabulary created during the model’s training. The resulting sequence of token IDs is what the model actually processes. Tokenization also defines the sequence length, which is the number of tokens in the input and is important because models have a maximum token limit.

#Overall, tokenization acts as a bridge between human-readable text and machine-readable numerical data, enabling language models to understand and process text effectively.
```

```python
# Install and import
# !pip install gensim
import gensim.downloader as api
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

print("--- C1: Word Embeddings ---")

# Load GloVe embeddings
print("Loading GloVe embeddings...")
glove_vectors = api.load("glove-wiki-gigaword-50")

# Select three words
words = ["king", "queen", "diamond"]

for word in words:
    if word in glove_vectors:
        print(f"\nWord: '{word}'")
        vector = glove_vectors[word]
        print(f"First 10 values of vector: {vector[:10]}")
        
        # Get most similar words
        similar_words = glove_vectors.most_similar(word, topn=5)
        print(f"Top 5 similar words:")
        for similar_word, similarity in similar_words:
            print(f"  {similar_word:15} → {similarity:.4f}")
    else:
        print(f"Word '{word}' not in vocabulary")
```

```python
#5.Sentence-Level Embedding
#Here we are calculating the cosine similarity between the statements by using the cosine similarity library provided by sklearn.
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

sentences = [
    "AI is changing the world",
    "Machine learning helps computers learn",
    "Deep neural networks process information",
    "Natural language understanding is complex",
    "Artificial intelligence transforms industries"
]

def embed(text):
    words = [glove_vectors[w] for w in text.lower().split() if w in glove_vectors]
    return np.mean(words, axis=0)

# Create sentence embeddings
vectors = [embed(s) for s in sentences]
sim = cosine_similarity(vectors)

# Find most similar pair (excluding self-similarity)
most_similar = None
best_score = -1

for i in range(len(sim)):
    for j in range(i+1, len(sim)):
        if sim[i][j] > best_score:
            best_score = sim[i][j]
            most_similar = (i, j)

print("\nSimilarity Matrix:\n", sim)
print(f"\nMost similar pair:\n'{sentences[most_similar[0]]}' ↔ '{sentences[most_similar[1]]}'")
print(f"Score: {best_score:.3f}")
```

```python
#6. Transformer Application
#Here we are using the Hugging face Transformer Pipleine to analyse the sentiment of the statements. This provides whether the statement is carrying negative sentiment or positive sentiment
```

```python
from transformers import pipeline
import pandas as pd
from tabulate import tabulate

print("=== Hugging Face Sentiment Classification ===\n")

# Initialize the sentiment analysis pipeline
print("Loading sentiment analysis model...")
classifier = pipeline("sentiment-analysis", 
                     model="distilbert-base-uncased-finetuned-sst-2-english")

# Custom inputs for sentiment analysis
custom_inputs = [
    "I absolutely love this product! It's amazing and works perfectly.",
    "This is the worst experience I've ever had. Terrible service.",
    "The weather is nice today.",
    "I feel okay about the decision, not great but not bad either.",
    "This movie is fantastic! The acting is superb and the plot is engaging.",
    "I'm extremely disappointed with the quality of this item.",
    "It's a decent product for the price, does what it needs to do.",
    "Outstanding performance by the team, they exceeded all expectations!",
    "The food was mediocre and overpriced, wouldn't recommend.",
    "I have mixed feelings about this - some aspects are good, others aren't."
]

print(f"Analyzing sentiment for {len(custom_inputs)} custom inputs...\n")

# Perform sentiment analysis
results = classifier(custom_inputs)
print(results)
```
